```python
import yfinance as yf
import pandas as pd

# ----------------------------
# 1️⃣ Download data
# ----------------------------
date_start = '2020-10-01'
date_end = '2025-11-01'

gold = yf.download("GC=F", start=date_start, end=date_end)['Close'] / 31.1035  # per gram
sp500 = yf.download("^GSPC", start=date_start, end=date_end)['Close']

# ----------------------------
# 2️⃣ Align indices and merge
# ----------------------------
gold.index = pd.to_datetime(gold.index)
sp500.index = pd.to_datetime(sp500.index)

df = pd.concat([gold, sp500], axis=1)
df.columns = ['Gold', 'SP500']
df = df.dropna()

# ----------------------------
# 3️⃣ Correlation of raw prices
# ----------------------------
corr_prices = df['Gold'].corr(df['SP500'])
print("Correlation of prices:", corr_prices)

# ----------------------------
# 4️⃣ Correlation of daily returns
# ----------------------------
df['Gold_ret'] = df['Gold'].pct_change()
df['SP500_ret'] = df['SP500'].pct_change()

corr_returns = df['Gold_ret'].corr(df['SP500_ret'])
print("Correlation of daily returns:", corr_returns)

# ----------------------------
# 5️⃣ Correlation on S&P down days only
# ----------------------------
down_days = df[df['SP500_ret'] < 0]
corr_down_days = down_days['Gold_ret'].corr(down_days['SP500_ret'])
print("Correlation on S&P down days:", corr_down_days)
```
