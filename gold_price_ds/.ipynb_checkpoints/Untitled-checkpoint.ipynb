```python
import yfinance as yf
import pandas as pd
import numpy as num


gold = yf.download("GC=F", start="2025-10-01", end="2025-10-10")
xrate = yf.download("USDIDR=X", start="2025-10-01", end="2025-10-10")
merge = gold.join(xrate,lsuffix= '_gold', rsuffix = '_xrate') 
price_col = ['Close', 'High', 'Low', 'Open']
merge_idr = pd.DataFrame()
for col in price_col
merge_idr[(col, 'IDR')] = merge[(col,'GC=F')] * merge[(col, 'USDIDR=X')]



#[price_col] = gold[price_col] * xrate["close"]

#gold.to_csv("gold.csv")
#gold = pd.read_csv("gold.csv")
#gold = gold.columns.values
print(merge.columns)
```
